````python
## add data
time = [1, 2, 3, 4, 5]
temperature = [25, 26, 28, 27, 30]

plt.plot(time, temperature) ## add x,y 
plt.title("Water Temperature Over Time") ## add title

## add label 
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")

## show graph 
plt.show()
````

### Customize Graph

````python
plt.figure(figsize = (8,4)) ## customize figure size

## customize line
plt.plot(
 time,
 temperature,
 color="blue",
 linestyle="-",
 marker="o"
)

## set Y limit
plt.ylim(20, 35)

## add gridlines
plt.grid(True)

## add title & label 
plt.title("Water Temperature Over Time")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.show()
````

### Error Bar Plotting

````python
plt.figure(figsize = (8,4)) ## customize figure size

temperature_error = [0.5, 0.8, 0.6, 0.7, 1.0]

## customize line
plt.errorbar(
 time,
 temperature,
 yerr=temperature_error,
 fmt="o-",
 color="blue",
 ecolor="blue",
 capsize=5,
 capthick=0.5
)

## set Y limit
plt.ylim(20, 35)

## add gridlines
plt.grid(True)

## add title & label 
plt.title("Water Temperature Over Time")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.show()
````

### Legend

````python
temperature_a = [25, 26, 28, 27, 30]
temperature_b = [24, 25, 26, 26, 28]
plt.plot(time, temperature_a, label="Station A")
plt.plot(time, temperature_b, label="Station B")
plt.title("Water Temperature at Two Stations")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.show()
````

### Combining Main Plot Components

````python
plt.figure(figsize=(9, 5))


## Station A plot
plt.plot(
 time,
 temperature_a,
 color="blue",
 linestyle="-",
 marker="o",
 label="Station A"
)

## Station B plot
plt.plot(
 time,
 temperature_b,
 color="red",
 linestyle="--",
 marker="s",
 label="Station B"
)

## customize graph 
plt.title("Water Temperature at Two Stations")
plt.xlabel("Time (hours)")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
````

## Lecture Part II : More Plotting Types

### Bar Chart 

````python
import matplotlib.pyplot as plt
stations = ["Station A", "Station B", "Station C", "Station D"]
turbidity = [5.2, 8.1, 3.7, 6.4]

plt.figure(figsize=(6, 4))

## bar 
bars = plt.bar(
 stations,
 turbidity,
 color="gold",
 edgecolor="black"
)

## bar label
plt.bar_label(bars, fmt="%.1f", padding=0)

plt.title("Turbidity Across Monitoring Stations")
plt.xlabel("Monitoring Station")
plt.ylabel("Turbidity (NTU)")

plt.show()
````

### Histogram

````python
import matplotlib.pyplot as plt
pm25 = [12, 15, 18, 20, 22, 22, 25, 27, 29, 31,
 33, 35, 36, 40, 42, 45, 48, 52, 60, 75]

plt.figure(figsize=(6, 4))

plt.hist(pm25, bins=5, edgecolor="black")

plt.title("Distribution of PM2.5 Concentrations")
plt.xlabel("PM2.5 Concentration (µg/m³)")
plt.ylabel("Frequency")
plt.show()
````

### Bins

````python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.hist(pm25, bins=5, edgecolor="black")
plt.title("Histogram with 5 Bins")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("Frequency")
plt.subplot(1, 2, 2)
plt.hist(pm25, bins=10, edgecolor="black")
plt.title("Histogram with 10 Bins")
plt.xlabel("PM2.5 (µg/m³)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
````

### Scatter plot

````python
plt.figure(figsize=(6, 4))

temperature = [20, 22, 24, 26, 28, 30, 32]
dissolved_oxygen = [9.1, 8.7, 8.2, 7.8, 7.3, 6.9, 6.4]

plt.scatter(
 temperature,
 dissolved_oxygen,
 color="teal",
 marker="o",
 s=80,
 alpha=0.7,
 edgecolor="black"
)
plt.title("Temperature vs. Dissolved Oxygen")
plt.xlabel("Water Temperature (°C)")
plt.ylabel("Dissolved Oxygen (mg/L)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

### Regression Line with Seaborn

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.DataFrame({
 "Temperature": [20, 22, 24, 26, 28, 30, 32],
 "DO": [8.9, 8.7, 8.5, 7.8, 7.0, 6.9, 6.6]
})

plt.figure(figsize=(8, 5))

## regression plot 
sns.regplot(
 data=df,
 x="Temperature",
 y="DO",
 ci = 95 , ## 95% confidence interval 
    scatter_kws={"color": "teal", "s": 80}, ## customize scatter 
    line_kws={"color": "red", "linewidth": 2} ## customize line
)


plt.title("Relationship Between Temperature and Dissolved Oxygen")
plt.xlabel("Water Temperature (°C)")
plt.ylabel("Dissolved Oxygen (mg/L)")
plt.tight_layout()
plt.show()
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

water_data = pd.DataFrame({
 "Temperature": [20, 22, 24, 26, 28, 30],
 "DO": [8.9, 8.7, 8.4, 7.8, 7.2, 6.9],
 "Conductivity": [102, 109, 109, 111, 116, 116],
 "Nitrate": [1.1, 1.6, 1.4, 1.7, 1.3, 1.4]
})

## define correlation map 
corr = water_data.corr(numeric_only=True)

## plot heatmap 
plt.figure(figsize=(5, 4))
sns.heatmap(
 corr,
 annot=True,
 cmap="coolwarm",
 vmin=-1,
 vmax=1
)

plt.title("Correlation Matrix of Water Quality Variables")
plt.tight_layout()
plt.show()
````

### Lab 3 : Exploratory Data Analysis (EDA)

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("river_water_quality.csv" )
print(df.head())
````

````python
print(df.shape)
print(df.columns)
print(df.info())
````

````python
missing_count = df.isnull().sum()
print(missing_count)

missing_percent = (df.isnull().mean() * 100).round(2)

print(missing_percent)
````

### Visualize Missing Data 

````python
missing_percent = (df.isnull().mean() * 100).sort_values(ascending=False) ## descending

plt.figure(figsize=(9, 5))

## bar plot
sns.barplot(
 x=missing_percent.values,
 y=missing_percent.index,
 color="steelblue"
)

## decorate graph 
plt.title("Percentage of Missing Values by Variable")
plt.xlabel("Missing Values (%)")
plt.ylabel("Variable")
plt.tight_layout()
plt.show()
````

### Convert the Timestamp to Datetime 

````python
df["Timestamp"] = pd.to_datetime(df["Timestamp"],errors="coerce") ## not a time (NaT) for coerce
print(df["Timestamp"].dtype)
print(df["Timestamp"].head())
````

### Task: Visualize the Distribution of Nitrate with Seaborn 

````python
plt.figure(figsize = (9,5))

### histogram plot with kde
sns.histplot( data = df , x = "NO3" , bins = 30 , kde = True ) ## bins = bar width 

### decorate graph 
plt.title("Distribution of Nitrate Measurements") 
plt.xlabel("Nitrate")
plt.ylabel("Frequency") 

plt.tight_layout()
plt.plot()
````

### Task: Visualize Nitrate Over Time 

````python
## makig time series first 
time_df = df.dropna(subset=["Timestamp", "NO3"]).sort_values("Timestamp")

plt.figure(figsize=(12, 5))

## line plot 
sns.lineplot(
 data=time_df,
 x="Timestamp",
 y="NO3",
 errorbar=None
)

## decorate graph
plt.title("Nitrate Measurements Over Time")
plt.xlabel("Timestamp")
plt.ylabel("Nitrate")
plt.tight_layout()
plt.show()
````

### Task: Examine Monthly Patterns 

````python
## groupby then aggregate and drop NA 
monthly_nitrate = (
 df.groupby("Month", dropna=False)["NO3"]
 .median()
 .dropna()
)

## bar plot 
plt.figure(figsize=(9, 5))
sns.barplot(
 x=monthly_nitrate.index, ### x as index of series 
 y=monthly_nitrate.values, ### y as values of series
 color="teal"
)

## decorate graph 
plt.title("Median Nitrate Concentration by Month")
plt.xlabel("Month")
plt.ylabel("Median Nitrate (Original Dataset Units)")
plt.tight_layout()
plt.show()
````

### Task : Interactive Scatter Matrix 

````python
## create column date and hour 

df["Date"] = df["Timestamp"].dt.date
df["Hour"] = df["Timestamp"].dt.hour

## change each column to numeric 
variables = ["Conductivity", "NO3", "Temp", "Turbidity"]
for col in variables:
    df[col] = pd.to_numeric(df[col], errors="coerce")
````

````python
plot_df = df[variables].dropna()

## plotting scatter matrix 
fig = px.scatter_matrix(
 plot_df,
 dimensions=variables,
 title="Pairwise Relationships Between Water Quality Variables",
 opacity=0.5
)

## update layout
fig.update_layout(
 width=1000,
 height=850
)

fig.show()
````
